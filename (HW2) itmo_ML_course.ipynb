# О наборе данных

Данные содержат 999 записей (рассмотрено ниже) с разными признаками, влияющими на выпадение волос: генетика, гормональные изменения, медицинские состояния, лечение, питание, стресс, возраст, вредные привычки и другие. Все факторы должны позволить модели учитывать комплекс факторов для наиболее точного предсказания. Датасет взят из каггла (линк: https://www.kaggle.com/datasets/amitvkulkarni/hair-health/data)

Описание каждого из полей (с каггла):

Genetics: Indicates whether the individual has a family history of baldness (Yes/No).

Hormonal Changes: Indicates whether the individual has experienced hormonal changes (Yes/No).

Medical Conditions: Lists specific medical conditions that may contribute to baldness, such as Alopecia Areata, Thyroid Problems, Scalp Infection, Psoriasis, Dermatitis, etc.

Medications & Treatments: Lists medications and treatments that may lead to hair loss, such as Chemotherapy, Heart Medication, Antidepressants, Steroids, etc.

Nutritional Deficiencies: Lists nutritional deficiencies that may contribute to hair loss, such as Iron deficiency, Vitamin D deficiency, Biotin deficiency, Omega-3 fatty acid deficiency, etc.

Stress: Indicates the stress level of the individual (Low/Moderate/High).

Age: Represents the age of the individual.

Poor Hair Care Habits: Indicates whether the individual practices poor hair care habits (Yes/No).

Environmental Factors: Indicates whether the individual is exposed to environmental factors that may contribute to hair loss (Yes/No).

Smoking: Indicates whether the individual smokes (Yes/No).

Weight Loss: Indicates whether the individual has experienced significant weight loss (Yes/No).

Baldness (Target): Binary variable indicating the presence (1) or absence (0) of baldness in the individual.

### Цель: разработать и оценить бейзлайн-модель для задачи бинарной классификации для предсказания облысения

### 1. Добавим недостающие импорты (в идеале - думать наперед и добавлять их в самом начале ноутбука, но наперед думать я не умею, так что просто придерживаюсь правила линейности воспроизведения ячеек и добавляю все что приходит в голову непосредственно перед сами использованием)

```python
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import recall_score, classification_report, confusion_matrix, roc_curve, auc
from sklearn.dummy import DummyClassifier
```

```python
# Зафиксируем рандом для получения одних и тех же рез-в
RANDOM_STATE = 2003
np.random.seed(RANDOM_STATE)
```

```python
print(df_forAnalisys.columns) # вспомним столбцы, чтобы не бегать далеко. Предобработки данные не требуют, с прошлого дз оставил все в 
# +- готовом для дальнейшего анализа виде
y = df_forAnalisys['Hair_Loss'] # целевая перем-я
```

```python
df_forAnalisys.info()
# df_forAnalisys.Medical_Conditions.unique()
# df_forAnalisys['Medications_&_Treatments'].unique()
df_forAnalisys['Nutritional_Deficiencies_'].unique()
```


### Избавлюсь полностью от OBJECT dtype, приведу их для начала к простому булевому значению: 0 - No Data (будем считать, что ничего человек не использует/не принимает/мед. спец условий у него нет). Иначе - 1. Вообще в идеале можно было бы их привести к какой-то прогрессивной истории, но спец. знаний в этой сфере у меня нет, буду надеяться на хорошее качество модели в рамках булевых значений)


```python
df_forAnalisys['Medical_Conditions_boolean'] = df_forAnalisys['Medical_Conditions'].apply(lambda x: 0 if x == 'No Data' else 1)
df_forAnalisys['Nutritional_Deficiencies_boolean'] = df_forAnalisys['Nutritional_Deficiencies_'].apply(lambda x: 0 if x == 'No Data' else 1)
df_forAnalisys['Medications_&_Treatments_boolean'] = df_forAnalisys['Medications_&_Treatments'].apply(lambda x: 0 if x == 'No Data' else 1)
```

### Разбиваю данные на train и test выборки. Использую стратификацию по целевой переменной для сохранения пропорций классов в обеих выборках, хотя данные у нас и изначально идеально сбалансированы

```python
X = df_forAnalisys.drop(['Id', 'Hair_Loss','Medical_Conditions','Medications_&_Treatments','Nutritional_Deficiencies_', 'Age_group'], axis=1) 
# выкидываем ненужный id и целевую пер-ю, а также обджект данные и age_group во избежание мультиколлинеарности и бОльшей точности (тк 
# очевидно что для нормализованной age переменной градация меньше и модель будет точнее рассматривать возраст нежели группу, 
# но это не точно)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, 
    test_size=0.3, 
    random_state=RANDOM_STATE, 
    stratify=y
)
```

### Можно посмотреть на то, что у нас вышло после разбития:

```python
print(f"Размер обучающей выборки: {X_train.shape}")
print(f"Размер тестовой выборки: {X_test.shape}")
print(f"Пропорция классов в обучающей выборке:\n{y_train.value_counts(normalize=True)}")
print(f"Пропорция классов в тестовой выборке:\n{y_test.value_counts(normalize=True)}")
```

### В целом выглядит достаточно приятно, ничего во время делания дз глаза не порезало
### Также все используемые переменные в датасете легко переводятся к числовым/булевым переменным (стресс вполне нормально лег в градаци 0, 1, ... и тд), а остальные переменные перевели в нули и единички, так что ничего сверхестественного в выбранном датасете (к сожалению или к частью) делать не придется: подходит обычный One-Hot Encoding

### Также стоит отметить, что в переменной Х есть 2 hеально int поля: возраст и уровень стресса. Их нужно нормализовать, чтобы не давать лишнего вклада

```python
# обозначим типы переменных для препроцессинга
numerical_features = ['Age','Stress']
categorical_features = X.drop(numerical_features, axis=1).columns.tolist()

# препроцессор для обработки признаков
preprocessor = ColumnTransformer(
    transformers=[('num', StandardScaler(), numerical_features),('cat', OneHotEncoder(handle_unknown='ignore', sparse_output=False),\
                                                                 categorical_features)],remainder='passthrough') # remainder не обяз., на всякий случай

X_train_processed = preprocessor.fit_transform(X_train)
```

```python
print(f"Размерность данных: {X_train_processed.shape}")
```

В качестве простой бейзлайн-модели выбираем логистическую регрессию, чтобы с чего-то начать
она базовая, тк хорошо работает для задач классификации

```python
baseline_model = Pipeline(steps=[('preprocessor', preprocessor),('classifier', \
                                                                 LogisticRegression(random_state=RANDOM_STATE, solver='liblinear', class_weight='balanced'))])
baseline_model.fit(X_train, y_train)
```

Модель обучилась, можно смотреть на метрики

```python
y_pred_baseline = baseline_model.predict(X_test)
recall_baseline = recall_score(y_test, y_pred_baseline, pos_label=1)
print(recall_baseline)
```

```python
confusion_matrix(y_test, y_pred_baseline)
```

### Что также интересно: OneHotEncoder в целом в моей выборке не нужен, тк я уже сам на нули и единички все категориальные столбцы разбил, но если в препроцессинг прям его включать, то он как раз улучшает score.

```python
y_proba = baseline_model.predict_proba(X_test)[:, 1]
fpr, tpr, thresholds = roc_curve(y_test, y_proba)
roc_auc = auc(fpr, tpr)
plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve')
plt.plot([0, 1], [0, 1], color='navy', lw=2, linestyle='--', label='Random Guess')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate / Recall')
plt.title('ROC curve')
plt.legend(loc="lower right")
plt.grid(True)
plt.show()

print(f"ROC AUC для бейзлайн-модели: {roc_auc:.4f}")
```
